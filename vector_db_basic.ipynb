```python
import chromadb

client = chromadb.Client()
collection = client.create_collection("news_article")

collection.add(
    ids = ["id1", "id2", "id3", "id4"],
    documents = [
        "HEADLINE: TechCorp Unveils Eco-Friendly 'Quantum-G' Battery for Smartphones. DATE: July 3, 2026. Silicon Valley electronics giant TechCorp announced a major breakthrough in mobile energy today with the release of its new 'Quantum-G' smartphone battery. The company claims this new hardware charges to 100% in just under five minutes and lasts for three full days on a single charge.",

        "Unlike traditional lithium-ion batteries that rely on scarce heavy metals, the Quantum-G is built entirely from organic, plant-based carbon synthetics. TechCorp CEO Elena Rostova stated during the keynote that this design reduces the manufacturing carbon footprint by 70% and makes the battery 100% recyclable at the end of its life cycle.",

        "Industry analysts are calling the announcement a massive win for environmental tech, though some urge caution. Market analyst Marcus Vance noted that while the laboratory test results are highly impressive, scaling mass production to meet global smartphone demands could take TechCorp at least 18 months.",

        "The company plans to pilot the new batteries in its flagship premium phone lineup starting in late 2027. Early developer kits are being shipped to software partners this week to optimize device power management systems."
    ]
)


```

```python
# Fetch documents + embeddings
data = collection.get(
    include=["documents", "embeddings", "metadatas"]
)

# Print embeddings
for i, emb in enumerate(data["embeddings"]):
    print(f"\nDocument {i+1}")
    print("Text:", data["documents"][i])
    print("Embedding length:", len(emb))
    print("First 10 values:", emb[:10])

```

```python
results = collection.query(
    query_texts=["this is the new battery"],
    n_results= 2
)
converted = results
print(converted)
```
